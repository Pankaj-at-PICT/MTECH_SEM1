```python
class SocialNetwork:

    def __init__(self):
        self.graph = {}

    def add_association(self, user1, user2):
        if user1 not in self.graph:
            self.graph[user1] = []

        if user2 not in self.graph:
            self.graph[user2] = []

        self.graph[user1].append(user2)
        self.graph[user2].append(user1)

    def get_association(self, user):
        return self.graph[user]

    def get_all_associations(self):
        return self.graph

    def get_no_of_associations(self, user):
        return len(self.graph[user])

    def get_total_associated_users(self):
        return len(self.graph)
```

```python
social_network = SocialNetwork()

def display_menu():
    print("\n--- Social Network Menu ---")
    print("1. Add Association")
    print("2. Get Associations for a User")
    print("3. Get All Associations")
    print("4. Get Number of Associations for a User")
    print("5. Get Total Associated Users")
    print("6. Exit")
    print("-------------------------")

while True:
    display_menu()
    choice = input("Enter your choice (1-6): ")

    if choice == '1':
        user1 = input("Enter first user: ")
        user2 = input("Enter second user: ")
        social_network.add_association(user1, user2)
        print(f"Association added between {user1} and {user2}.")
    elif choice == '2':
        user = input("Enter user to get associations for: ")
        try:
            associations = social_network.get_association(user)
            print(f"Associations for {user}: {associations}")
        except KeyError:
            print(f"User '{user}' not found in the network.")
    elif choice == '3':
        all_associations = social_network.get_all_associations()
        print("All Associations:")
        if all_associations:
            for user, connections in all_associations.items():
                print(f"  {user}: {connections}")
        else:
            print("  No associations yet.")
    elif choice == '4':
        user = input("Enter user to get number of associations for: ")
        try:
            count = social_network.get_no_of_associations(user)
            print(f"Number of associations for {user}: {count}")
        except KeyError:
            print(f"User '{user}' not found in the network.")
    elif choice == '5':
        total_users = social_network.get_total_associated_users()
        print(f"Total number of associated users: {total_users}")
    elif choice == '6':
        print("Exiting social network menu. Goodbye!")
        break
    else:
        print("Invalid choice. Please enter a number between 1 and 6.")
```
