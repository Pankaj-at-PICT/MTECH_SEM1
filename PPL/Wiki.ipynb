```python
import requests
import pandas as pd

url = "https://en.wikipedia.org/wiki/List_of_Wikipedias"

headers = {
    "User-Agent": "Mozilla/5.0"
}

tables = pd.read_html(url,storage_options=headers)

```

```python
print("Number of tables:", len(tables))
```

```python
for i, table in enumerate(tables):
    print(f"\nTable {i}")
    print(table.columns)
```
